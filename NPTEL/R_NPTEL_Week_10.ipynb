## Section 1: Pattern Matching and Replacement

### Example 1: Basic Substitution with `sub()`

```r
y <- "Number of participants: 25"
result <- sub("25", "30", y)
print(result)
```

### Example 2: First-Occurrence Limitation of `sub()`

```r
y <- "Mr. Singh is the smart one. Mr. Singh is funny, too."
print(y)
result <- sub("Mr. Singh", "Professor Jha", y)
print(result)
```

### Example 3: Global Substitution with `gsub()`

```r
y <- "Mr. Singh is the smart one. Mr. Singh is funny, too."
result <- gsub("Mr. Singh", "Professor Jha", y)
print(result)
```

### Comparison of `sub()` and `gsub()` behavior

```r
sub("Mr. Singh", "Professor Jha", y)
gsub("Mr. Singh", "Professor Jha", y)
```

### Example 4: Value Extraction using `grep()`

```r
str <- c(
  "R Course",
  "exercises",
  "include examples of R language"
)
result <- grep("ex", str, value = TRUE)
print(result)
```

### Example 5: Finding Indices with `grep()`

```r
str <- c(
  "R Course",
  "exercises",
  "include examples of R language"
)
result <- grep("ex", str, value = FALSE)
print(result)
```

### Example 6: Case-Insensitive Matching with `grep()`

```r
str <- c(
  "R Course",
  "exercises",
  "include examples of r language",
  "in R software."
)
result1 <- grep(
  "R",
  str,
  ignore.case = FALSE,
  value = TRUE
)
print(result1)
result2 <- grep(
  "R",
  str,
  ignore.case = TRUE,
  value = TRUE
)
print(result2)
```

### Example 7: Index Retrieval & Case Sensitivity Options

```r
str <- c(
  "R Course",
  "exercises",
  "include examples of r language",
  "in R software."
)
grep(
  "R",
  str,
  ignore.case = TRUE,
  value = FALSE
)
grep(
  "R",
  str,
  ignore.case = FALSE,
  value = FALSE
)
```

### Example 8: Searching within Vector Constructs

```r
x <- "R course 24.07.2021"
y <- "Number of participants: 25"
combined <- c(x, y)
print(combined)
result <- grep("our", combined)
print(result)
```

### Example 9: Matching Target Prefix "Num"

```r
x <- "R course 24.07.2022"
y <- "Number of participants: 50"
combined <- c(x, y)
print(combined)
result <- grep("Num", combined)
print(result)
```

### Example 10: Logical Vector Creation via `grepl()`

```r
str <- c(
  "R Course",
  "exercises",
  "include examples of R language"
)
result <- grepl("R", str)
print(result)
```

### Example 11: Boolean Checks for Specific Patterns

```r
str <- c(
  "R Course",
  "exercises",
  "include examples of R language"
)
result <- grepl("ex", str)
print(result)
```
