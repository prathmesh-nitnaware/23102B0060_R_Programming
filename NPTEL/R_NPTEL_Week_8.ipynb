```r
# Red = 1
# Blue = 2
# Green = 3

colours = c("Red", "Green", "Green", "Blue", "Red")

colours
```

```r
x = c("Red", "Green", "Green", "Blue", "Red")

f = factor(x)

f
```

```r
y = c(1, 4, 3, 5, 4, 2, 4)

possible.dieface = c(1, 2, 3, 4, 5, 6)

labels.dieface = c(
  "one", "two", "three",
  "four", "five", "six"
)

facy = factor(
  y,
  levels = possible.dieface,
  labels = labels.dieface
)

facy
```
