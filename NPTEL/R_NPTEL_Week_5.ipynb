## Missing Data

### NA and is.na()

```r
x <- c(1, 2, NA, 4, 5)
is.na(x)
```

### Missing Data in Vectors

```r
y <- c(10, NA, 30, NA, 50)
is.na(y)
```

### mean() with Missing Values

```r
mean(x)
mean(x, na.rm = TRUE)
```

### NA versus NULL

```r
is.na(NULL)
is.null(NA)
length(NA)
length(NULL)
```

### which(is.na())

```r
z <- c(1, NA, 3, 4, NA, 6)
which(is.na(z))
```

### Counting Missing Values

```r
sum(is.na(z))
```

### complete.cases()

```r
df <- data.frame(A = c(1, 2, NA, 4), B = c('a', NA, 'c', 'd'))
complete.cases(df)
df[complete.cases(df), ]
```

### na.omit()

```r
na.omit(df)
```

## Handling Missing Values

### if()

```r
a <- 10
if (a > 5) {
  print("a is greater than 5")
}
```

### if-else

```r
b <- 3
if (b > 5) {
  print("b is greater than 5")
} else {
  print("b is not greater than 5")
}
```

### Logical Operators && and ||

```r
x_cond <- 7
y_cond <- 12

if (x_cond > 5 && y_cond < 15) {
  print("Both conditions are true")
}

if (x_cond < 5 || y_cond < 10) {
  print("At least one condition is true")
} else {
  print("Neither condition is true")
}
```

### Nested if-else if

```r
score <- 85

if (score >= 90) {
  print("Grade A")
} else if (score >= 80) {
  print("Grade B")
} else if (score >= 70) {
  print("Grade C")
} else {
  print("Grade F")
}
```

### ifelse()

```r
grades <- c(75, 92, 60, 88)
result <- ifelse(grades >= 70, "Pass", "Fail")
print(result)
```

### Vectorized Conditional Execution

```r
numbers <- 1:10
category <- ifelse(numbers %% 2 == 0, "Even", "Odd")
print(category)
```

### Even and Odd Numbers with ifelse()

```r
my_numbers <- c(1, 4, 7, 2, 9, 6)
odd_even <- ifelse(my_numbers %% 2 == 0, "Even", "Odd")
print(odd_even)
```

### switch()

```r
choice <- 2
switch(choice, "one", "two", "three")
```

### switch() with Numeric Expressions

```r
day_num <- 3
switch(day_num, "Monday", "Tuesday", "Wednesday", "Thursday", "Friday")
```

### switch() with Character Expressions

```r
fruit <- "apple"
switch(fruit, apple = "Red fruit", banana = "Yellow fruit", orange = "Orange fruit")
```

### switch() with No Match

```r
color <- "purple"
switch(color, red = 1, blue = 2, green = 3, "Unknown color")
```

### which()

```r
vals <- c(10, 20, 5, 30, 15)
which(vals > 15)
```

### which() with Logical Conditions

```r
data_vec <- c(T, F, T, T, F)
which(data_vec == T)
```

### which.min()

```r
numbers_min <- c(5, 1, 8, 2)
which.min(numbers_min)
```

### which.max()

```r
numbers_max <- c(5, 1, 8, 2)
which.max(numbers_max)
```

### which() with Matrices

```r
mat <- matrix(1:9, nrow = 3)
which(mat %% 2 == 0)
```

### which() with arr.ind=TRUE

```r
which(mat %% 2 == 0, arr.ind = TRUE)
```
