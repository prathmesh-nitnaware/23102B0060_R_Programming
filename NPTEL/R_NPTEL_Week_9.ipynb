## `cat()` Function for Concatenation and Printing

```r
cat("The zero occurs at", 2*pi, "radians.", "\n")
```

```r
d = date()
cat("Today's date is:", d, "\n")
```

```r
x = 1:10
x

cat(x, sep = " ++ ")
cat("\n")

cat(x, sep = " / ")
```

```r
x = 7

cat("The square of", x, "is", x^2, "!\n")
```

```r
x = 7

cat(
  "The square root of", x,
  "is approximately",
  format(sqrt(x), digits = 3),
  "\n"
)
```

```r
evenno = c(2, 4, 6, 8, 10)

evenno

cat("The first few even numbers are:", evenno, "...\n")
```

## `paste()` and `paste0()` Functions for Concatenating Vectors

```r
paste(
  "Everybody",
  "loves",
  "R Programming.",
  sep = "*"
)
```

```r
paste(
  "Everybody",
  "loves",
  "R Programming.",
  sep = "==="
)
```

```r
paste(1:12)
```

```r
as.character(1:12)
```

```r
names = c(
  "Prof. Singh",
  "Mr. Venkat",
  "Dr. Jha"
)

names

paste(
  names,
  "is",
  "a good",
  "person."
)
```

```r
names = c(
  "Prof. Singh",
  "Mr. Venkat",
  "Dr. Jha"
)

paste(
  names,
  "is",
  "a good",
  "person.",
  collapse = ", and "
)
```

## `strsplit()` Function for Splitting Strings

```r
x = "The&!syntax&!of&!paste&!is!&available!&inthe online-help"

strsplit(x, split = "&!")
```

```r
dates = c(
  "2020-07-24",
  "2021-08-25",
  "2022-09-26",
  "2023-10-27"
)

dates
```

```r
datesplt = strsplit(dates, "-")

datesplt
```

```r
datemat = matrix(
  unlist(datesplt),
  nrow = 4,
  ncol = 3,
  byrow = TRUE
)

datemat
```

```r
datematrix = matrix(
  as.numeric(unlist(datesplt)),
  nrow = 4,
  ncol = 3,
  byrow = TRUE
)

datematrix
```

## `nchar()` and `nzchar()` Functions for Character Counts

```r
x = c("Apple", "Banana", "Cake")

nchar(x)
```

```r
y = c(2, 4, 6)

nchar(y)
```

```r
z = c(11, 222, 3333)

nchar(z)
```

```r
z1 = c(1.1, 2.22, 3.333)

nchar(z1)
```

```r
x = "R course 24.07.2022"
y = "Number of participants: 25"

nzchar(x)
nzchar(y)
```

```r
x = c("Apple", "Banana", "Cake")

x

nzchar(x)
```

```r
y = c("Apple", "", "Cake")

y

nzchar(y)
```

```r
x = "R course will start from 24.07.2022"

x

toupper(x)
```
