## Nested `for` Loops

```r
x = c(2, 4, 6, 8, 10, 12)

excount = function(x) {
  count = 0

  for (xval in x) {
    if (xval / 2 > 3)
      count = count + 1
  }

  print(count)
}

excount(x)
```

```r
child = c("child1", "child2", "child3")
sweet = c("sweet1", "sweet2", "sweet3")

for (x in child) {
  for (y in sweet) {
    print(paste(x, y))
  }
}
```

## Loop Control: `break` Statement

```r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "tea") {
    break
  }

  print(x)
}
```

## Loop Control: `next` Statement

```r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "lemonade") {
    next
  }

  print(x)
}
```

## Basic `while` Loop

```r
i = 1

while (i < 10) {
  print(i^2)
  i = i + 2
}
```

## Interactive `while` Loop with a Function

```r
sumfunction = function() {
  sum = 0

  number = as.integer(
    readline(prompt = "Please select any number less than 25: ")
  )

  while (number <= 25) {
    sum = sum + number
    number = number + 1
  }

  print(paste(
    "The sum of numbers received from the While Loop: ",
    sum
  ))
}
```

```r
sumfunction()
```

## Basic `repeat` Loop

```r
22 + 23 + 24 + 25
```

```r
i = 1

repeat {
  print(i^2)

  i = i + 2

  if (i > 10)
    break
}
```

## `repeat` Loop with `next` and `break`

```r
i = 1

repeat {
  i = i + 1

  if (i < 10)
    next

  print(i^2)

  if (i >= 13)
    break
}
```

# Functions in R
## Defining and Calling a Function with One Argument

```r
abc = function(x) {
  x^2
}
```

```r
abc(3)
```

```r
abc(6)
```

```r
abc(9)
```

## Defining and Calling a Function with Multiple Arguments

```r
abc = function(x, y) {
  x^2 + y^2
}
```

```r
abc(3, 4)
```

```r
abc(10, 10)
```

```r
abc(-2, -3)
```

## Redefining a Function with Trigonometric Operations

```r
abc = function(x) {
  sin(x)^2 + cos(x)^2 + x
}
```

```r
abc(9)
```

```r
abc(99)
```

```r
abc(-15)
```

## Defining and Calling a Function with No Arguments

```r
abc = function() {
  for (i in 1:3) {
    print(i^3)
  }
}
```

```r
abc()
```
