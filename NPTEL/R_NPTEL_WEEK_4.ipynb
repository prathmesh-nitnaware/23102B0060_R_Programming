## Logical Operators
### Basic Logical Operations (AND, OR, NOT)

```r
x <- TRUE
y <- FALSE

x & y
x | y
!x
```

### Short-Circuiting OR (||)

```r
x <- 8

(x < 10) || (x < 2)

x <- 18

(x < 10) || (x < 2)
```

### Element-wise OR (|) with Vectors

```r
x <- c(8, 18)

x1 <- 8

(x1 < 10) || (x1 < 2)

(x < 10) | (x < 2)
```

### Short-Circuiting AND (&&) vs. Element-wise AND (&)

```r
x <- c(8, 18)

# && is used with a single value
x1 <- 8

(x1 < 10) && (x1 > 2)

# & is used element-by-element with a vector
(x < 10) & (x > 2)
```

### Using Logical Operators for Subsetting (AND)

```r
x <- 1:6

# Check which values satisfy both conditions
(x > 2) & (x < 5)

# Select only those values
x[(x > 2) & (x < 5)]
```

### Using Logical Operators for Subsetting (OR)

```r
x <- 1:6

# Check which values satisfy either condition
(x > 2) | (x < 5)

# Select those values
x[(x > 2) | (x < 5)]
```

### Exclusive OR (xor)

```r
x <- TRUE
y <- FALSE

xor(x, y)
```

### Checking Truth Values (isTRUE, isFALSE)

```r
isTRUE(8 < 6)

isTRUE(8 > 6)

isFALSE(5 < 8)

isFALSE(5 > 8)
```

## Logical Variables
### Creating Logical Variables

```r
x <- 5

Logical1 <- (x > 2)
Logical1
is.logical(Logical1)

Logical2 <- (x < 10)
Logical2
is.logical(Logical2)

Logical3 <- (x != 5)
Logical3
is.logical(Logical3)
```

### Further Examples of Logical Variables

```r
x <- 5

Logical4 <- (2 * x > 11)
Logical4
is.logical(Logical4)

Logical5 <- (3 * x < 20)
Logical5
is.logical(Logical5)
```
