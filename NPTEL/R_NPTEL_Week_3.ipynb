## 1. Power Operations

```r
print('--- Power Operations Examples ---')

# Using ^ operator
print(paste('2^3:', 2^3))
print(paste('2^0.5:', 2^0.5))
print(paste('2^-0.5:', 2^-0.5))

# Using ** operator
print(paste('2**3:', 2**3))
print(paste('2**0.5:', 2**0.5))

# Vector power operations
print(paste('c(2,3,5,7)^2:', c(2,3,5,7)^2))
print(paste('c(2,3,5,7)^c(2,3):', c(2,3,5,7)^c(2,3))) # Recycling c(2,3) to match length 4
print(paste('c(1,2,3,4,5,6)^c(2,3,4):', c(1,2,3,4,5,6)^c(2,3,4))) # Recycling c(2,3,4) to match length 6
print(paste('c(2,3,5,7)^c(2,3,4):', c(2,3,5,7)^c(2,3,4))) # Recycling c(2,3,4) to match length 4
```

## 2. Integer Division

```r
print('--- Integer Division Examples (%%/%) ---')

# Scalar integer division
print(paste('2 %/% 2:', 2 %/% 2))
print(paste('5 %/% 2:', 5 %/% 2))
print(paste('7 %/% 3:', 7 %/% 3))

# Vector integer division
print(paste('c(2,3,5,7) %/% 2:', c(2,3,5,7) %/% 2))
print(paste('c(2,3,5,7) %/% c(2,3):', c(2,3,5,7) %/% c(2,3))) # Recycling c(2,3)
print(paste('c(2,3,5) %/% c(2,3):', c(2,3,5) %/% c(2,3))) # Recycling c(2,3), warning due to unequal lengths
```

### Practice Task 2: Integer Division
Perform integer division of the vector `c(10, 15, 22, 30)` by `c(3, 4)`.

```r
# Solution to Practice Task 2
vec1 <- c(10, 15, 22, 30)
vec2 <- c(3, 4)
result <- vec1 %/% vec2
print(result)
```

## 3. Modulo Division

```r
print('--- Modulo Division Examples (%%) ---')

# Scalar modulo division
print(paste('2 %% 2:', 2 %% 2))
print(paste('3 %% 2:', 3 %% 2))
print(paste('7 %% 3:', 7 %% 3))
print(paste('7 %% 4:', 7 %% 4))

# Vector modulo division
print(paste('c(2,3,5,7) %% 2:', c(2,3,5,7) %% 2))
print(paste('c(2,3,5,7) %% c(2,3):', c(2,3,5,7) %% c(2,3))) # Recycling c(2,3)
print(paste('c(2,3,5) %% c(2,3):', c(2,3,5) %% c(2,3))) # Recycling c(2,3), warning due to unequal lengths
```

### Practice Task 3: Modulo Division

```r
# Solution to Practice Task 3
vec1 <- c(11, 16, 23, 31)
vec2 <- c(3, 5)
result <- vec1 %% vec2
print(result)
```

## 4. Built-in Functions

```r
print('--- Built-in Functions Examples ---')

# max(), min()
print(paste('max(5, 10, 2):', max(5, 10, 2))) # Multiple scalar arguments
print(paste('max(c(5, 10, 2)):', max(c(5, 10, 2)))) # Vector argument
print(paste('min(5, 10, 2):', min(5, 10, 2)))
print(paste('min(c(5, 10, 2)):', min(c(5, 10, 2))))

# mean()
print(paste('mean(c(1,2,3,4,5)):', mean(c(1,2,3,4,5))))

# abs()
print(paste('abs(-5):', abs(-5)))
print(paste('abs(c(-1, 0, 1)):', abs(c(-1, 0, 1))))

# sqrt()
print(paste('sqrt(9):', sqrt(9)))
print(paste('sqrt(c(4, 9, 16)):', sqrt(c(4, 9, 16))))

# sum()
print(paste('sum(c(1,2,3)):', sum(c(1,2,3))))

# prod()
print(paste('prod(c(1,2,3)):', prod(c(1,2,3))))

# round()
print(paste('round(3.14159, 2):', round(3.14159, 2)))
print(paste('round(c(3.1, 3.5, 3.9)):', round(c(3.1, 3.5, 3.9))))

# floor()
print(paste('floor(3.7):', floor(3.7)))
print(paste('floor(c(3.1, 3.9, -2.5)):', floor(c(3.1, 3.9, -2.5))))

# ceiling()
print(paste('ceiling(3.1):', ceiling(3.1)))
print(paste('ceiling(c(3.1, 3.9, -2.5)):', ceiling(c(3.1, 3.9, -2.5))))

# log(), log10(), log2(), exp()
print(paste('log(10):', log(10))) # Natural logarithm
print(paste('log10(100):', log10(100)))
print(paste('log2(8):', log2(8)))
print(paste('exp(1):', exp(1))) # e^1

# Trigonometric functions (angles in radians)
print(paste('sin(pi/2):', sin(pi/2)))
print(paste('cos(pi):', cos(pi)))
print(paste('tan(pi/4):', tan(pi/4)))

# Inverse trigonometric functions (returns radians)
print(paste('asin(1):', asin(1)))
print(paste('acos(0):', acos(0)))
print(paste('atan(1):', atan(1)))

# Hyperbolic functions
print(paste('sinh(1):', sinh(1)))
print(paste('cosh(0):', cosh(0)))
print(paste('tanh(0):', tanh(0)))

# Inverse hyperbolic functions
print(paste('asinh(0):', asinh(0)))
print(paste('acosh(1):', acosh(1)))
print(paste('atanh(0):', atanh(0)))
```

## 5. Assignments

```r
print('--- Assignment Examples ---')

# Assignment using =
x1 = c(1,2,3,4)
print(paste('x1:', paste(x1, collapse = ', ')))

# Assignment using <-
x2 <- x1^2
print(paste('x2 (x1 squared):', paste(x2, collapse = ', ')))

# Calculations with assignments
# Demonstrate complex calculation
vector_op <- c(1,2,3,4)
sum_vec <- sum(vector_op)
prod_vec <- prod(c(1,2))

result1 <- vector_op + sum_vec * prod_vec
print(paste('c(1,2,3,4) + sum(c(1,2,3,4)) * prod(c(1,2)):', paste(result1, collapse = ', ')))

result2 <- abs(vector_op - sum_vec * prod_vec)
print(paste('abs(c(1,2,3,4) - sum(c(1,2,3,4))*prod(c(1,2))):', paste(result2, collapse = ', ')))
```

## 6. Matrices

```r
print('--- Matrix Creation Examples ---')

# Default matrix creation (column-wise filling)
mat_default <- matrix(nrow=4, ncol=2, data=c(1,2,3,4,5,6,7,8))
print('Matrix (default, column-wise):')
print(mat_default)

# Explicitly column-wise filling
mat_colwise <- matrix(nrow=4, ncol=2, data=c(1,2,3,4,5,6,7,8), byrow=FALSE)
print('Matrix (byrow=FALSE, column-wise):')
print(mat_colwise)

# Row-wise filling
mat_rowwise <- matrix(nrow=4, ncol=2, data=c(1,2,3,4,5,6,7,8), byrow=TRUE)
print('Matrix (byrow=TRUE, row-wise):')
print(mat_rowwise)

# Demonstrating default fill direction
cat('\nDefault matrix fill direction is column-wise, which means data fills the first column, then the second, and so on.\n')
cat('When `byrow=TRUE`, data fills the first row, then the second, and so on.\n')
```

## 7. Matrix Properties

```r
print('--- Matrix Properties Examples ---')

# Create a sample matrix 'x'
x <- matrix(c(10, 20, 30, 40, 50, 60, 70, 80), nrow = 4, ncol = 2, byrow = TRUE)
print('Sample matrix x:')
print(x)

# Accessing elements
print(paste('x[3,2] (element at row 3, column 2):', x[3,2]))

# Dimensions
print(paste('dim(x) (dimensions of x):', paste(dim(x), collapse = 'x')))
print(paste('nrow(x) (number of rows):', nrow(x)))
print(paste('ncol(x) (number of columns):', ncol(x)))

# Mode and Attributes
print(paste('mode(x) (storage mode of x):', mode(x))) # Data type of elements
print('attributes(x) (attributes of x):')
print(attributes(x))

# Type checking and conversion
print(paste('is.matrix(x) (is x a matrix?):', is.matrix(x)))
# For demonstration, converting back to matrix from a similar structure if it weren't a matrix
# as.matrix(x) would return x itself if it's already a matrix
# Example: if y was a data frame, as.matrix(y) would convert it.
# Let's create a data frame for as.matrix demo
df_example <- data.frame(A = 1:2, B = 3:4)
print('\nExample with as.matrix on a data.frame:')
print(df_example)
print(as.matrix(df_example))
```

## 8. Matrix Help and Related Functions

```r
print('--- Matrix Help and Related Functions Examples ---')

# Accessing help documentation for matrix
# help("matrix") # Uncomment and run to see the help page
cat('Run `help("matrix")` in a separate cell to view the documentation for the matrix function.\n')

# Demonstrating matrix parameters: data, nrow, ncol, byrow, dimnames
# Example with dimnames
matrix_with_names <- matrix(data = 1:6, nrow = 2, ncol = 3,
                          dimnames = list(c('Row1', 'Row2'), c('ColA', 'ColB', 'ColC')))
print('\nMatrix with dimnames:')
print(matrix_with_names)

# Demonstrating recycling when data contains too few elements
# The data vector is `1:5` (5 elements), but the matrix is 3x3 (9 elements).
# R will recycle `1:5` to fill the remaining 4 positions (1, 2, 3, 4).
matrix_recycling <- matrix(data = 1:5, nrow = 3, ncol = 3)
print('\nMatrix demonstrating data recycling (data = 1:5, matrix is 3x3):')
print(matrix_recycling)
cat('Notice how the numbers 1 to 5 are repeated (recycled) to fill the 3x3 matrix.\n')

# Another example of recycling with warning
matrix_recycling_warning <- matrix(data = 1:4, nrow = 3, ncol = 3) # 4 elements for 9 spots, not a multiple
print('\nMatrix demonstrating data recycling with warning (data = 1:4, matrix is 3x3):')
print(matrix_recycling_warning)
cat('A warning is typically issued when the length of the data is not a multiple of the number of elements needed for the matrix.\n')
```

## 9. Matrix vs Data Frame

```r
print('--- Matrix vs Data Frame Examples ---')

# Load the warpbreaks dataset
data(warpbreaks)
print('\nFirst few rows of warpbreaks data:')
print(head(warpbreaks))

# Check if warpbreaks is a matrix
print(paste('is.matrix(warpbreaks) (Is warpbreaks a matrix?):', is.matrix(warpbreaks)))

# Display first 10 rows of warpbreaks (it's a data.frame)
print('\nwarpbreaks[1:10,] (first 10 rows):')
print(warpbreaks[1:10,])

# Convert first 10 rows to a matrix and display
as_matrix_warpbreaks <- as.matrix(warpbreaks[1:10,])
print('\nas.matrix(warpbreaks[1:10,]) (first 10 rows converted to matrix):')
print(as_matrix_warpbreaks)

# Check the type after conversion
print(paste('is.matrix(as_matrix_warpbreaks) (Is the converted object a matrix?):', is.matrix(as_matrix_warpbreaks)))
```

### Practice Task 9: Matrix vs Data Frame
Create a small data frame with two columns: `name` (character) and `age` (numeric). Then, try to convert it into a matrix and observe the result, especially the data type.

```r
# Solution to Practice Task 9
# Create a data frame
my_df <- data.frame(name = c('Alice', 'Bob', 'Charlie'), age = c(25, 30, 22))
print('Original Data Frame:')
print(my_df)
print(paste('Mode of Data Frame:', mode(my_df)))
print(paste('Is my_df a data.frame?', is.data.frame(my_df)))

# Convert to matrix
my_matrix_from_df <- as.matrix(my_df)
print('\nConverted to Matrix:')
print(my_matrix_from_df)
print(paste('Mode of converted Matrix:', mode(my_matrix_from_df)))
print(paste('Is my_matrix_from_df a matrix?', is.matrix(my_matrix_from_df)))
cat('\nNotice how the numeric `age` column became character after conversion to a matrix, because all elements in a matrix must be of the same type.')
```
